```r
# ---- CONFIGURACION ----
notebook_base <- "Prueba_624_WorkFlow_01_junior_grupoA.ipynb"   # <- poné el nombre real
prefijo_exp   <- "9400"

#semillas <- c(102191)   # probá primero con una
semillas <- c(999917, 999907, 999883, 999863, 999853) 
               616111, 664211, 687121, 884951, 915851)

            
n <- length(semillas)

carpeta <- paste0("corridas_", prefijo_exp)
dir.create(carpeta, showWarnings = FALSE)

# ---- ARMADO DEL SCRIPT ----
armar_bloque <- function(i) {
  s <- semillas[i]
  siguiente <- if (i < n) paste0("siguiente: semilla ", semillas[i + 1])
               else "era la ultima"
  c(
    paste0('echo "[$(date +%H:%M:%S)] INICIO corrida ', i, ' de ', n,
           ' | semilla=', s, '"'),
        paste0("SEMILLA=", s, " EXPERIMENTO=", prefijo_exp, "_s", s,
           " jupyter nbconvert --to notebook --execute",
           " --ExecutePreprocessor.timeout=-1",
           " --ExecutePreprocessor.kernel_name=ir ",
           shQuote(notebook_base),
           " --output-dir ", shQuote(carpeta),
           " --output ", prefijo_exp, "_s", s, ".ipynb"),
    paste0('echo "[$(date +%H:%M:%S)] FIN corrida ', i, ' de ', n,
           ' | semilla=', s, ' | codigo de salida=$? | ', siguiente, '"'),
    'echo ""'
  )
}

lineas <- c(unlist(lapply(seq_len(n), armar_bloque)),
            'echo "[$(date +%H:%M:%S)] TODAS LAS CORRIDAS TERMINARON"')

archivo_sh  <- file.path(carpeta, "run_loop.sh")
archivo_log <- file.path(carpeta, "log_loop.txt")
writeLines(lineas, archivo_sh)

# ---- LANZAMIENTO EN SEGUNDO PLANO ----
system(paste("nohup bash", shQuote(archivo_sh), ">", shQuote(archivo_log), "2>&1"), wait = FALSE)
cat("Lanzado. Mirá el log con la celda de abajo.\n")
```
